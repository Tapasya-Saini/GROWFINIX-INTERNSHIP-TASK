```python
import pandas as pd
import matplotlib.pyplot as plt

# Teri wali excel ka data direct yahi daal diya hai
data = {
    'video_group': ['A','A','A','A','A','B','B','B','B','B'],
    'CTR': [0.12, 0.13, 0.115, 0.125, 0.118, 0.18, 0.19, 0.185, 0.178, 0.192]
}
df = pd.DataFrame(data)
avg = df.groupby('video_group')['CTR'].mean()*100

plt.bar(['Video A', 'Video B'], avg, color=['red', 'green'])
plt.ylabel('Average CTR %')
plt.title('CTR Comparison - ab_test_data se')
plt.show()

print(avg)
```
